## Import Toolkit

```python
from pymycobot.mycobot280 import MyCobot280
from pymycobot import PI_PORT, PI_BAUD
import time
```

## Connect to the arm

```python
mc = MyCobot280(PI_PORT, PI_BAUD)
```

## Set RGB color

```python
# Red
mc.set_color(255, 0, 0)
```

```python
# Blue
mc.set_color(0, 0, 255)
```

```python
# Green
mc.set_color(0, 255, 0)
```

```python
# White
mc.set_color(255, 255, 255)
```

```python
# Orange
mc.set_color(225, 103, 36)
```

## Demo：LED color loop

```python
i = 7 

while i > 0:                            
    time.sleep(1.5)         
    mc.set_color(255, 0, 0) #Red
    time.sleep(2)
    mc.set_color(0, 255, 0) #green
    time.sleep(3)
    mc.set_color(0, 0, 255) #blue
    i -= 1
```
